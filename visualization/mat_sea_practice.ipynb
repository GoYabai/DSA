# Load dataset

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")

print(df.head())
```

# bài 1: sns.displot hay distribution plot

```python
g = sns.displot(df, x="total_bill", hue="time", kind="hist", kde=True, height=5, aspect=1.5)

g.ax.set_title("Phân phối Giá trị Hóa đơn theo Bữa ăn", fontsize=14, fontweight='bold')
g.ax.set_xlabel("Tổng hóa đơn ($)")
g.ax.set_ylabel("Số lượng khách")
g.ax.set_xlim(0, 60)
plt.show()


```

# Bài 2: sns.relplot hay relation plot

```python
g = sns.relplot(df, x="total_bill", y="tip", size="size", col="smoker", alpha=0.7, height=5, aspect=1.2)

g.fig.suptitle("Mối quan hệ giũa chúng ta là gì")

for ax in g.axes.flat:
    ax.set_xlabel("Tổng hóa đơn ($)")
    ax.set_ylabel("Tiền bo ($)")

plt.show()
```

# Bài 3: sns.catplot hay cate plot

```python
g = sns.catplot(data=df, x="day", y="tip", hue="sex", kind="bar", height=5, aspect=1.2)

ax = g.ax
ax.set_title("Trung bình Tiền bo theo Ngày và Giới tính", fontsize=14, pad=20)
ax.set_xlabel("Ngày trong tuần")
ax.set_ylabel("Tiền bo trung bình")

from matplotlib.ticker import FuncFormatter
ax.yaxis.set_major_formatter(FuncFormatter(lambda y, _: f'${y:g}'))

ax.yaxis.grid(True, linestyle='--', alpha=0.7)


plt.show()
```
